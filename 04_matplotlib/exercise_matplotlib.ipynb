### Task 6 — Combined 2×2 subplot figure: combine histogram, bar chart, box plot and scatter plot in a single figure.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].hist(df['power_kw'],
        bins=40,
        color='steelblue',
        edgecolor='white')
axes[0, 0].set_xlabel('Power (kW)')
axes[0, 0].set_ylabel('Count')
axes[0, 0].set_title('Distribution of Charging Power (kW)')

top10 = df['city'].value_counts().head(10)   # value_counts(): zählt die Häufigkeit und sortiert diese Werte absteigend
axes[0, 1].barh(top10.index[::-1],    # Muster [start:stop:step], : beginnt automatisch am richtigen Ort (Start und Ende)
        top10.values[::-1],
        color='steelblue',
        edgecolor='white')
axes[0, 1].set_xlabel('Number of Charging Stations')
axes[0, 1].set_title('Top 10 Cities by Number of EV Charging Stations')

order = ['AC_1_PHASE', 'AC_3_PHASE', 'DC']
data = [df[df['power_type'] == t]['power_kw'].values for t in order]
axes[1, 0].boxplot(data, tick_labels=order, patch_artist=True,
                   boxprops=dict(facecolor='steelblue', alpha=0.6))
axes[1, 0].set_xlabel('Power Type')
axes[1, 0].set_ylabel('Power (kW)')
axes[1, 0].set_title('Charging Power by Power Type')

for ptype, grp in df.groupby('power_type'):
    axes[1, 1].scatter(grp['voltage'], grp['amperage'], label=ptype, alpha=0.5, s=30)

axes[1, 1].set_xlabel('Voltage (V)')
axes[1, 1].set_ylabel('Amperage (A)')
axes[1, 1].set_title('Voltage vs. Amperage by Power Type')
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```
