### Task 1 — Histogram of `power_kw`

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(df['power_kw'],
        bins=40,
        color='steelblue',
        edgecolor='white')
ax.set_xlabel('Power (kW)')
ax.set_ylabel('Count')
ax.set_title('Distribution of Charging Power (kW)')
plt.tight_layout()
plt.show()
```
